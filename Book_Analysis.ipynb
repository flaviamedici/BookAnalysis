```python
with open("miracle_in_the_andes.txt", "r", encoding="utf-8") as file:
    book = file.read()
```

### How many chapters?

#### With string method

```python
import re
```

```python
pattern = re.compile("Chapter [0-9]+")
pattern
```

```python
re.findall(pattern, book)
```
